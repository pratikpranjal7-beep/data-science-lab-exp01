### Question 1
Strings in Python are **immutable**, meaning their value cannot be changed in place after creation. Therefore, string methods like `.lower()` and `.upper()` construct and return a brand new string reflecting the transformation.

### Question 4

```python
word = 'Hello'
print('word[1:3]:', word[1:3])
print('word[-4:-1]:', word[-4:-1])
```

**Explanation:**
Negative indices count backward from the end of the string (`-1` is the last character). Slicing `[-4:-1]` starts at index `-4` (`'e'`) up to, but not including, index `-1` (`'o'`).

### Question 5

```python
word_num = '42'
print('int:', int(word_num), type(int(word_num)))
print('float:', float(word_num), type(float(word_num)))

# Calling int('Hi') raises a ValueError
try:
    word = 'Hi'
    int(word)
except ValueError as e:
    print('Calling int("Hi") raises:', e)
```

## 4. Dictionaries in Python

### Question 16

```python
d = {'name': 'Alice', 'age': 25}
print('Keys:', list(d.keys()))
print('Values:', list(d.values()))
print('Retrieving age:', d['age'])
```

### Question 17

```python
d = {'a': 1}
print('dict.get("b"):', d.get('b'))
try:
    d['b']
except KeyError as e:
    print('dict["b"] raises KeyError:', e)
```

**Explanation:** `dict.get(key)` is safer because it returns `None` (or a default value) instead of halting execution with a `KeyError`.

### Question 18
Modern Python (3.7+) dictionaries preserve insertion order. Order matters when sequentially processing keys/values or outputting ordered configurations.

### Question 19

```python
d = {'a': 1, 'b': 2, 'c': 3}

# Safe pattern: create a list of keys to remove first
keys_to_remove = [k for k, v in d.items() if v > 1]
for k in keys_to_remove:
    d.pop(k)

print('Modified dictionary:', d)
```

### Question 20

```python
person = {'name': 'Alex', 'role': 'Developer', 'level': 3}
for key, value in person.items():
    print(f'{key}: {value}')
```

## 5. List and Dictionary Comprehension

### Question 21

```python
result = [i**2 for i in range(1,11)]
print('Result:', result)
```

**Explanation:**
- `i**2`: Expression calculating the square of `i`.
- `for i in range(1,11)`: Loop generating numbers from 1 to 10.

### Question 22

```python
result = [i for i in range(6)]
print(result)
```

### Question 23

```python
evens = [i for i in range(0, 19, 2)]
print(evens)
```

### Question 24

```python
squares_dict = {i: i**2 for i in range(1, 11)}
print(squares_dict)
```

## 6. Sets

### Question 26
1. **Unordered:** Sets do not maintain positional index order.
2. **Unique elements:** Duplicate elements are ignored.
3. **Unindexed collection of immutable/hashable items.**

### Question 27

```python
colors = set(['blue', 'green', 'red'])
colors.add('blue')
print('Colors set:', colors)
```

**Explanation:** No change occurs because sets enforce element uniqueness.

### Question 28

```python
s = {1, 2, 3}
s.discard(99)
print('s.discard(99) completed with no error.')

try:
    s.remove(99)
except KeyError as e:
    print('s.remove(99) raises KeyError:', e)
```

### Question 29

```python
set1 = {'blue', 'green', 'red'}
set2 = {'blue', 'yellow'}
print('Union:', set1.union(set2))
print('Intersection:', set1.intersection(set2))
```

### Question 30

```python
set1 = {'blue', 'green', 'red'}
set2 = {'blue', 'yellow'}
print('Union (|):', set1 | set2)
print('Intersection (&):', set1 & set2)
```
