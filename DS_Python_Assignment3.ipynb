# Section B – Medium (10 Questions)

```python
# Q16
# Output will return labeled entries for 'Q' (45) and 'S' (61).
s = Series([18, 45, 27, 61], index=['P', 'Q', 'R', 'S'])
print(s[s > 30])
```

```python
# Q17
# 'West' in s  -> True  (checks index labels)
# 500 in s     -> False (checks index labels, not values)
s = Series([500, 600, 700], index=['East', 'West', 'North'])
print('West' in s)
print(500 in s)
```

```python
# Q18
# Label 'Chennai' receives NaN because it is not in city_data.
city_data = {'Delhi': 80, 'Pune': 65, 'Kochi': 70}
s = Series(city_data, index=['Delhi', 'Kochi', 'Pune', 'Chennai'])
print(s)
```

```python
# Q19
# Conceptual pd.isnull(s) output:
# W: False, X: False, Y: False, Z: True
s = Series({'W': 5, 'X': 10, 'Y': 15}, index=['W', 'X', 'Y', 'Z'])
print(pd.isnull(s))
```

```python
# Q20
# Conceptual pd.notnull(s) output:
# W: True, X: True, Y: True, Z: False
print(pd.notnull(s))
```

```python
# Q21
# Result adds values aligned by matching index labels:
# A: 10 + 2 = 12
# B: 20 + 3 = 23
# C: 30 + 1 = 31
s1 = Series([10, 20, 30], index=['A', 'B', 'C'])
s2 = Series([1, 2, 3], index=['C', 'A', 'B'])
print(s1 + s2)
```

```python
# Q22
# Pandas aligns data by index labels during operations.
# 'A' exists only in s1 and 'C' exists only in s2, so performing s1 + s2 results in NaN for labels 'A' and 'C'.
s1 = Series({'A': 12, 'B': 18})
s2 = Series({'B': 5, 'C': 9})
print(s1 + s2)
```

```python
# Q23
stock = Series([25, 40, 18], index=['Item1', 'Item2', 'Item3'])
stock_dict = stock.to_dict()
stock2 = Series(stock_dict)
print(stock_dict)
print(stock2)
```

# Section C – Hard (5 Questions)

```python
# Q26
# Pandas aligns on index labels:
# 'Jan': 40 + NaN = NaN
# 'Feb': 55 + 20 = 75.0
# 'Mar': 60 + 25 = 85.0
# 'Apr': NaN + 30 = NaN
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south
print(total)
```

```python
# Q27
# (a) 'Yellow' contains NaN.
# (b) pd.isnull(result) returns True for 'Yellow'.
# (c) It is created because 'Yellow' is present in labels but missing in dictionary d.
# (d) 'Green' (24.0), 'Red' (12.0), and 'Blue' (18.0) retain actual numeric values.
d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green', 'Red', 'Yellow', 'Blue']
result = Series(d, index=labels)
print(result)
print(pd.isnull(result))
```

```python
# Q28
# s2 index: ['Ravi', 'Aman', 'Neha'] | values: [90, 75, 82]
# Yes, s and s2 represent the same labelled data.
s = Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = Series(d)
print(s2)
print(s.equals(s2))
```

```python
# Q29
prod_sales = Series([450, 600, 300, 800], index=['P1', 'P2', 'P3', 'P4'])
print("Values:", prod_sales.values)
print("Index:", prod_sales.index)
print("Select P2:", prod_sales['P2'])
print("Sales > 500:", prod_sales[prod_sales > 500])
print("P1 exists:", 'P1' in prod_sales)
dict_sales = prod_sales.to_dict()
print("Dictionary:", dict_sales)
```

```python
# Q30
# (a) Index order: ['C', 'A', 'D', 'B']
# (b) Label 'D' contains NaN
# (c) s[s > 20] returns 'C': 25.0 and 'B': 35.0
# (d) pd.notnull(s): C: True, A: True, D: False, B: True
# (e) Series name: 'Observations', Index name: 'Category'
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print(s)
print(s[s > 20])
print(pd.notnull(s))
```
