# Section C - Hard (5 Questions)

```python
# Q26
course_data = {
    'Student': ['Ann', 'Ben', 'Cara', 'Dan', 'Elsa'],
    'Course': ['DS', 'AI', 'ML', 'DS', 'AI'],
    'Score': [90, 85, 88, 92, 95]
}
course_df = pd.DataFrame(course_data)
print(course_df.columns)
print(course_df['Student'])
student_score_df = course_df[['Student', 'Score']]
print(student_score_df)
course_df['Batch'] = '2026'
print(course_df)
```

```python
# Q27
travel_df = pd.DataFrame({'Destination': ['Paris', 'London', 'Tokyo', 'Rome', 'Berlin']})
travel_df['Code'] = np.arange(5)
print("After assigning np.arange(5):")
print(travel_df)
travel_df['Code'] = pd.Series(['T1', 'T4'], index=[1, 4])
print("After replacing with partial Series:")
print(travel_df)
```

```python
# Q28
school_data = {
    'Student': ['A', 'B', 'C'],
    'Class': [10, 10, 10],
    'Section': ['A', 'B', 'A']
}
school_df = pd.DataFrame(school_data)
school_df_extended = pd.DataFrame(school_df, columns=['Student', 'Class', 'Section', 'House'])
school_df_extended['House'] = 'Gryffindor'
print(school_df_extended)
```

```python
# Q29
inventory_data = {
    'Item': ['Pen', 'Pencil', 'Eraser', 'Ruler'],
    'Qty': [100, 200, 150, 80]
}
inventory_df = pd.DataFrame(inventory_data)
inventory_df['Status'] = pd.Series(['In Stock', 'In Stock'], index=[0, 2])
print("After adding partial Series:")
print(inventory_df)
del inventory_df['Status']
print("After deleting added column:")
print(inventory_df)
```

```python
# Q30
lab_df = pd.read_clipboard()
print("1. Full DataFrame:", lab_df)
print("2. Column names:", lab_df.columns)
single_word_col = lab_df.columns[0]
print("3. Single-word column:", lab_df[single_word_col])
space_col = [col for col in lab_df.columns if ' ' in col]
if space_col:
    print("4. Column with space:", lab_df[space_col[0]])
smaller_df = lab_df[lab_df.columns[:2]]
print("5. Smaller DataFrame:", smaller_df)
lab_df['TempCol'] = 'Temporary'
print("6. After adding TempCol:", lab_df)
del lab_df['TempCol']
print("7. After deleting TempCol:", lab_df)
```
