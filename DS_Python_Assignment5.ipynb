# Section B - Moderate (20 Questions)

## NumPy

```python
# Q21
sales = np.array([
    [120, 150, 180],
    [140, 160, 175],
    [135, 155, 190],
    [150, 170, 200]
])
daily_sales = np.sum(sales, axis=1)
product_sales = np.sum(sales, axis=0)
highest_day = np.argmax(daily_sales) + 1
print("Total sales per day:", daily_sales)
print("Total sales per product:", product_sales)
print("Day with highest sales:", f"Day {highest_day}")
```

```python
# Q22
grid = np.arange(10, 46).reshape(6, 6)
submatrix = grid[2:6, 1:5]
sub_mean = np.mean(submatrix)
print("Submatrix:")
print(submatrix)
print("Submatrix Mean:", sub_mean)
```

```python
# Q23
weights = np.array([48, 62, 75, 81, 54, 69, 90])
above_70 = weights[weights > 70]
modified_weights = np.where(weights < 55, 55, weights)
print("Weights above 70:", above_70)
print("Modified weights:", modified_weights)
print("Original weights preserved:", weights)
```

```python
# Q24
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])
item_costs = quantity * price
total_bill = np.sum(item_costs)
pct_contrib = (item_costs / total_bill) * 100
print("Each product total cost:", item_costs)
print("Overall bill:", total_bill)
print("Percentage contributions (%):", pct_contrib)
```

```python
# Q25
arr = np.array([
    [12, 15, 18, 21],
    [24, 27, 30, 33],
    [36, 39, 42, 45]
])
row_var = np.var(arr, axis=1)
row_std = np.std(arr, axis=1)
col_var = np.var(arr, axis=0)
col_std = np.std(arr, axis=0)
print("Row Variance:", row_var)
print("Row Std Dev:", row_std)
print("Column Variance:", col_var)
print("Column Std Dev:", col_std)
# Explanation:
# axis=0 calculates stats down columns across rows.
# axis=1 calculates stats across rows through columns.
```

```python
# Q26
mat1 = np.array([[1, 2, 3], [4, 5, 6]])  # 2x3
mat2 = np.array([[7, 8], [9, 10], [11, 12]])  # 3x2
matmul_result = np.dot(mat1, mat2)
print("Matrix Multiplication Result (2x2):")
print(matmul_result)
# Explanation:
# Element-wise multiplication requires identical shapes (e.g. both 2x3).
# Matrix multiplication requires inner dimensions to match (2x3 @ 3x2 -> 2x2).
```

```python
# Q27
np.random.seed(42)
random_ints = np.random.randint(10, 100, size=25)
sorted_ints = np.sort(random_ints)
top_5 = sorted_ints[-5:]
unique_count = len(np.unique(random_ints))
print("Sorted integers:", sorted_ints)
print("5 largest values:", top_5)
print("Unique observations count:", unique_count)
```

```python
# Q28
a = np.array([5, 10, 15, 20, 25, 30])
b = a[1:5]
c = a[1:5].copy()
b[1] = 99
c[2] = 77
print("a:", a)
print("b:", b)
print("c:", c)
# Explanation:
# b is a view (shares same memory buffer as a), so b[1] = 99 modifies a.
# c is an independent deep copy, so changing c[2] does not affect a.
```

## Pandas Series

```python
# Q29
dept_a = pd.Series({'Jan': 250, 'Feb': 310, 'Mar': 290})
dept_b = pd.Series({'Mar': 180, 'Jan': 210, 'Apr': 260})
total_rev = dept_a + dept_b
print(total_rev)
# Labels producing missing values (NaN): 'Feb' (missing in dept_b) and 'Apr' (missing in dept_a).
# Explanation: Pandas aligns index labels; if a label isn't present in both Series, addition produces NaN.
```
