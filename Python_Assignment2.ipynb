# Topic 3: Objects, Identity (is) vs Equality (==)

```python
# Q3.1
# The '==' operator checks value equality (whether two objects hold equivalent contents).
# The 'is' operator checks object identity (whether two variables refer to the exact same location in memory).
```

```python
# Q3.2
# Prediction:
# a == b -> True (equal values)
# a is b -> False (distinct list objects in memory)
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

```python
# Q3.3
x = 100
y = 100
print(x is y, x == y)

p = 1000
q = 1000
print(p is q, p == q)

# Explanation: Python pre-allocates and caches small integers (typically from -5 to 256) in memory to optimize performance.
# Thus, x and y point to the same cached object, whereas p and q are created as separate objects.
```
